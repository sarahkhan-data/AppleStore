```python
from csv import reader
```

```python
AppleStore = list(reader(open("AppleStore.csv", encoding = "utf-8")))
```

```python
AppleStore[0].index("prime_genre")
```

```python
genre = {}
for i in AppleStore[1:]:
    prime= i[11]
    if prime in genre:
        genre[prime] += 1
    else:
        genre[prime] = 1
```

*generate a frequency table prime genre for free apps

```python
genre = {}
for i in AppleStore[1:]:
    prime= i[11]
    price = (float(i[4]))
    if price == 0:
        if prime in genre:
            genre[prime] += 1
        else:
            genre[prime] = 1
```

```python
genre
```

*generate a frequency table prime genre for paid apps

```python
genre = {}
for i in AppleStore[1:]:
    prime= i[11]
    price = (float(i[4]))
    if price != 0:
        if prime in genre:
            genre[prime] += 1
        else:
            genre[prime] = 1
```

```python
genre
```

*generate a frequency table prime genre for paid apps where user ratinngis greater than 4

```python
genre = {}
for i in AppleStore[1:]:
    prime= i[11]
    price = (float(i[4]))
    rating = (float(i[7]))
    if price != 0 and  rating > 4:
        if prime in genre:
            genre[prime] += 1
        else:
            genre[prime] = 1
```
